## Bag Of Words

The Bag of Words (BoW) model is a fundamental Natural Language Processing (NLP) technique that converts text   
into numerical vectors based on word frequency, ignoring grammar and word order.

**Key Aspects of Bag of Words**:  
**Representation**: Text is converted into vectors, commonly used in machine learning for text classification and information retrieval.  
**Methodology**: It involves tokenizing text, building a vocabulary, and creating a document-term matrix where each row is a document and   columns are word counts.  
**Limitations**: It loses context, semantic meaning, and word order. The resulting vectors can be sparse, especially with large vocabularies.  
**Preprocessing**: Often involves converting text to lowercase, removing punctuation, and sometimes removing stopwords.  
**Variations**: TF-IDF (Term Frequency-Inverse Document Frequency) is often used to weigh common words less, improving upon simple frequency counting.

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer()
```

```python
documents = [
    "cat sat mat",
    "dog sat log",
    "cat chased dog",
    "dogs are pets",
    "cat sat again"
]
```

```python
bow = cv.fit_transform(documents)
```

```python
vocab = cv.vocabulary_

# sort by index
sorted_vocab = sorted(vocab.items(), key=lambda x: x[1])

for word, idx in sorted_vocab:
    print(idx, word, end=' ')
```

```python
for b in bow:
    print(b.toarray())
```

```python
cv.transform(["cat sat mat"]).toarray()
```

**Pros of Bag of Words**  
**Simplicity & Interpretability**: Easy to understand and implement for beginners.  
**Computational Efficiency**: Minimal preprocessing (tokenization/cleaning) makes it fast for processing large text corpora.  
**Good Baseline**: Highly effective for basic text classification, document classification, and simple retrieval tasks.  
**Versatility**: Works well with many traditional machine learning algorithms (e.g., Naive Bayes, Logistic Regression). 

**Cons of Bag of Words**  
**Loss of Word Order & Context**: "Dog eats food" and "Food eats dog" are treated as identical because only counts matter, not sequence.  
 **High Dimensionality & Sparsity**: Large vocabularies create massive, mostly empty (sparse) vectors, increasing storage and   computational costs.  
**No Semantic Meaning**: Ignores synonyms; treats "car" and "vehicle" as entirely unrelated terms.  
**No Grammar/Negation Handling**: Cannot distinguish between "good" and "not good".  
**Out-of-Vocabulary (OOV)**: Cannot handle new words not seen in the training data.
