## TF IDF

TF-IDF (Term Frequency-Inverse Document Frequency) is a numerical statistic used in NLP to weigh the importance of a word within a document relative to a larger corpus. It highlights unique, meaningful words by multiplying how often a word appears in a document (TF) by the inverse frequency of the word across all documents (IDF).

**Term Frequency (TF)**  
Measures how frequently a word occurs in a document.  
TF = Number of timers a word t appears in document d / Total words in document d

**Inverse Term Frequency**    
Reduces the weight of common words and increases the weight of rare words.  
IDF = log( Total number of documents N / Number of documents containing word t)  
TF-IDF Formula = TF * IDF  
  
IDF will be higher if word is rare in documents. IDF will be lower if word  is coming in most of the documents.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
import pandas as pd
```

```python
d0 = 'The cat sits'
d1 = 'The dog runs'

string = [d0, d1]
```

```python
tfidf = TfidfVectorizer()
result = tfidf.fit_transform(string)
```

```python
print(tfidf.idf_)
feature_names = tfidf.get_feature_names_out()
print(feature_names)
```

```python
print('\nidf values:')
for ele1, ele2 in zip(tfidf.get_feature_names_out(), tfidf.idf_):
    print(ele1, ':', ele2)
```

```python
df = pd.DataFrame(result.toarray(), columns=feature_names, index=["Doc 1", "Doc 2"])
```

```python
print(df)
```
