## N Grams

N-grams in NLP are contiguous sequences of \(n\) items (words, characters, or tokens) from a text, used to analyze, predict, and model language patterns by capturing local context. Commonly used as Unigrams (\(n=1\)), Bigrams (\(n=2\)), or Trigrams (\(n=3\)), they are vital for tasks like predictive text, spell-checking, and machine translation.

**Usage Examples in NLP**  
**Predictive Text/Autocomplete**: Suggesting the next word based on previous words (e.g., in "I love...", the model predicts "natural" based on the bigram "love natural").  
**Spelling Correction**: Identifying unlikely sequences (e.g., "graet" vs. "great") based on n-gram frequency.  
**Text Classification/Sentiment Analysis**: Identifying phrases like "not good" (bigram) to accurately determine sentiment, which a single-word model might misinterpret.  
**Machine Translation**: Evaluating translation quality by comparing n-gram overlaps between machine and human translation

```python
def generate_ngrams(text, n):
    words = text.split()
    ngrams = []

    for i in range(len(words) - n + 1):
        gram = tuple(words[i:i+n])
        ngrams.append(gram)

    return ngrams


# Example
text = "I love natural language processing"
print(generate_ngrams(text, 2))
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(ngram_range=(1, 2))
```

```python
documents = [
    "cat sat mat",
    "dog sat log",
    "cat chased dog",
    "dogs are pets",
    "cat sat again"
]
```

```python
bow = cv.fit_transform(documents)
```

```python
vocab = cv.vocabulary_

# sort by index
sorted_vocab = sorted(vocab.items(), key=lambda x: x[1])

for word, idx in sorted_vocab:
    print(idx, word, end=' ')
```

```python
for b in bow:
    print(b.toarray())
```

```python
from nltk.util import ngrams

sentence = "I love NLP"
tokens = sentence.split()

result = list(ngrams(tokens, 2))
print(result)
```

**Pros of N-grams**  
**Simplicity & Speed**: They are easy to implement and computationally lightweight, making them ideal for rapid prototyping and low-resource systems.  
**Captures Local Context**: Unlike a standard "bag of words," n-grams preserve immediate word order. This helps models understand phrases like "not good" or "New York City" that lose meaning when split into single words.  
**Effective Baseline**: They provide a strong starting point for tasks like text classification, sentiment analysis, and plagiarism detection.  
**Robust to Noise**: Character-level n-grams can be resistant to spelling errors and typos, which is especially useful for analyzing social media text.  
**Language Agnostic**: They can be applied to new languages without needing extensive linguistic rules.

**Cons of N-grams**  
**Data Sparsity**: As the sequence length (\(n\)) increases, the likelihood of finding that specific sequence in training data drops significantly, leading to "zero probability" errors for unseen phrases.  
**Fixed Context Window**: N-grams only "see" a few words back. They cannot link words at the beginning of a long sentence to those at the end, missing long-range dependencies.  
**Exponential Growth**: Increasing the value of \(n\) causes the number of possible combinations to grow exponentially, requiring massive amounts of memory and storage.  
**Lack of Deep Meaning**: N-grams rely on exact matches and raw frequency; they don't understand synonyms or the abstract semantic relationship between words (e.g., they won't know "buy" and "purchase" are related unless explicitly seen together).
