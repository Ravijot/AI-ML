# Regularization in Machine Learning

## 🌿 What is Regularization?
**Regularization** is a technique used to prevent a model from learning noise in training data. It does this by penalizing complexity and encouraging simpler models that generalize better to unseen data.

**Formula**
Loss_new = Loss_original + λ × Penalty

- Loss_original → training error  
- λ (lambda) → regularization strength  
- Penalty → function of model parameters  

---

## 🎯 Why Do We Need It?
Without regularization, models may:
- Memorize training data
- Overfit
- Become overly complex

Regularization helps:
- Reduce overfitting
- Improve generalization
- Stabilize predictions

---

## 🧠 Types of Regularization

### 1. L1 Regularization (Lasso)
Penalty:
Σ |w_i|

**Effects**
- Some weights become zero
- Performs feature selection

**Use Case**
- Many irrelevant features

---

### 2. L2 Regularization (Ridge)
Penalty:
Σ (w_i²)

**Effects**
- Shrinks weights toward zero
- Keeps all features

**Use Case**
- Most features are useful

---

### 3. Elastic Net
Combination of L1 + L2

Penalty:
λ1 Σ |w_i| + λ2 Σ (w_i²)

**Effects**
- Balanced sparsity and stability

---

## ⚙️ Intuition
Without regularization → model fits every point including noise  
With regularization → model fits main pattern only  

---

## 📊 Choosing λ
- Small λ → complex model → risk of overfitting  
- Large λ → simple model → risk of underfitting  

Usually selected using **cross-validation**.

---

## 🏁 Comparison Table

| Type | Penalty | Effect |
|------|--------|-------|
| L1 | absolute weights | sparse model |
| L2 | squared weights | smooth model |
| Elastic Net | both | balanced |

---

## 🔑 Key Idea
**Regularization = penalty added to loss to control model complexity.**
