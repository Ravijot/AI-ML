# 🎯 Accuracy in Machine Learning

## 📌 What is Accuracy?

Accuracy measures how often a machine learning model makes the correct prediction.

### Formula (General Form)

Accuracy = (Number of Correct Predictions) / (Total Number of Predictions)

---

### Formula (Using Confusion Matrix Terms)

Accuracy = (TP + TN) / (TP + TN + FP + FN)

Where:

- TP = True Positive  
- TN = True Negative  
- FP = False Positive  
- FN = False Negative  

---

## 📦 Example

Suppose:

TP = 40  
TN = 45  
FP = 5  
FN = 10  

Accuracy = (40 + 45) / (40 + 45 + 5 + 10)  
Accuracy = 85 / 100  
Accuracy = 0.85 = 85%

---

## ✅ Advantages

- Easy to understand  
- Simple to compute  
- Works well for balanced datasets  

---

## ❌ Disadvantages

- Can be misleading for imbalanced datasets  

Example:  
If 95% of data belongs to one class, predicting only that class gives 95% accuracy — even if the model fails on the minority class.

---

## 🚦 When to Use Accuracy?

✔ When classes are balanced  
✔ When false positives and false negatives are equally important  

Avoid using only accuracy for:
- Fraud detection  
- Medical diagnosis  
- Spam detection  

In such cases, use Precision, Recall, F1-score, or ROC-AUC.

# 🎯 Cases Where Accuracy Fails

Accuracy looks clean and confident.  
But sometimes it behaves like a student who memorized only the easy questions.

Here are the major scenarios where accuracy becomes misleading:

---

## 1️⃣ Imbalanced Dataset (Most Common Failure Case)

When one class dominates the dataset.

### Example: Fraud Detection

- 1000 transactions  
- 980 are NOT fraud  
- 20 are fraud  

If a model predicts **"Not Fraud" for everything**:

Accuracy = 980 / 1000 = **98%**

Sounds amazing.  
But it failed to detect **all fraud cases**.

🚨 In such cases, use:
- Precision
- Recall
- F1-score
- ROC-AUC

---

## 2️⃣ Medical Diagnosis

Imagine:

- 99 healthy patients  
- 1 patient has cancer  

Model predicts everyone as healthy.

Accuracy = 99%

But it missed the **one critical case**.

Here, **Recall (Sensitivity)** matters much more than accuracy.

---

## 3️⃣ Unequal Cost of Errors

When false positives and false negatives have different consequences.

### Example: Spam Detection

- False Positive → Important email goes to spam  
- False Negative → Spam reaches inbox  

Accuracy treats both mistakes equally.  
Real-world impact does not.

---

## 4️⃣ Multi-Class with Rare Classes

If one class is very rare, accuracy may ignore it completely.

The model performs well on common classes  
and quietly fails on rare but important ones.

---

## 5️⃣ Highly Skewed Business Problems

Example:
- Predicting machine failure
- Cyber attack detection
- Disease outbreak prediction

Failures are rare but extremely important.

Accuracy rewards majority guessing.
Business reality does not.

---

# 🔎 Why Accuracy Fails

Because:

Accuracy = (Total Correct Predictions) / (Total Predictions)

It only counts how many are correct.  
It does not care *which* ones are wrong.

---

# ✅ Better Alternatives (When Accuracy Fails)

- **Precision** → How many predicted positives are correct?
- **Recall** → How many actual positives were detected?
- **F1-score** → Balance between precision and recall
- **ROC-AUC** → Model’s ability across thresholds

---

# 🏁 Final Thought

Accuracy works beautifully  
when classes are balanced  
and error costs are equal.

Outside that safe zone,  
it can be dangerously optimistic.

# 📊 Confusion Matrix in Machine Learning

A **Confusion Matrix** is a table used to evaluate the performance of a classification model.

It tells you not just *how many* predictions were correct,  
but *what kind* of mistakes were made.

---

# 🎯 Binary Classification Confusion Matrix

|                    | Predicted Positive | Predicted Negative |
|--------------------|-------------------|-------------------|
| **Actual Positive** | True Positive (TP) | False Negative (FN) |
| **Actual Negative** | False Positive (FP) | True Negative (TN) |

---

# 🧠 Meaning of Each Term

### ✅ True Positive (TP)
Model predicted **Positive**, and it is actually **Positive**.

### ✅ True Negative (TN)
Model predicted **Negative**, and it is actually **Negative**.

### ❌ False Positive (FP)
Model predicted **Positive**, but it is actually **Negative**.  
Also called **Type I Error**.

### ❌ False Negative (FN)
Model predicted **Negative**, but it is actually **Positive**.  
Also called **Type II Error**.

---

# 📦 Example

Suppose we are predicting disease:

- TP = 40  
- TN = 50  
- FP = 5  
- FN = 10  

Total samples = 105

---

# 📈 Metrics Derived from Confusion Matrix

### Accuracy
Accuracy = (TP + TN) / (TP + TN + FP + FN)

### Precision
Precision = TP / (TP + FP)

### Recall (Sensitivity)
Recall = TP / (TP + FN)

### Specificity
Specificity = TN / (TN + FP)

### F1 Score
F1 = 2 × (Precision × Recall) / (Precision + Recall)

---

# 🎯 Why Confusion Matrix is Important

Accuracy gives only one number.

Confusion matrix shows:
- What type of mistakes the model makes
- Whether it misses positives
- Whether it raises false alarms
- How balanced the predictions are

It is the foundation from which almost all classification metrics are calculated.

---

# 🏁 Summary

Confusion Matrix is the full report card.  
Accuracy is just the final percentage.

If you want to truly understand your model,  
always look at the confusion matrix first.

# 🎯 Precision and Recall in Machine Learning

When your model predicts "Positive", two important questions arise:

- 📌 How often is it correct?
- 📌 How many real positives did it actually find?

That is where **Precision** and **Recall** enter the stage.

---

# 🔵 Precision

## 📌 Definition

Precision measures how many predicted positives are actually correct.

Precision = TP / (TP + FP)

Where:
- TP = True Positive  
- FP = False Positive  

## 🧠 Intuition

Out of everything the model labeled as **Positive**,  
how many were truly positive?

It measures **quality of positive predictions**.

## 📦 Example

If:
- TP = 40  
- FP = 10  

Precision = 40 / (40 + 10)  
Precision = 40 / 50 = 0.80 = 80%

That means 80% of predicted positives were correct.

---

# 🟢 Recall (Sensitivity)

## 📌 Definition

Recall measures how many actual positives were correctly identified.

Recall = TP / (TP + FN)

Where:
- TP = True Positive  
- FN = False Negative  

## 🧠 Intuition

Out of all real positives,  
how many did the model successfully detect?

It measures **coverage of positive class**.

## 📦 Example

If:
- TP = 40  
- FN = 20  

Recall = 40 / (40 + 20)  
Recall = 40 / 60 = 0.67 = 67%

The model detected 67% of actual positives.

---

# ⚖️ When to Use Precision?

Use Precision when **false positives are costly**.

### Examples:
- Spam detection (important email marked as spam)
- Loan approval (risky person approved)
- Search engine results

Here, you care about being correct when you say "Positive".

---

# 🚨 When to Use Recall?

Use Recall when **false negatives are costly**.

### Examples:
- Disease detection
- Fraud detection
- Security threats

Here, you care about not missing real positive cases.

---

# 🎯 Precision vs Recall Tradeoff

Usually:

- Increasing Precision decreases Recall  
- Increasing Recall decreases Precision  

It is like tightening or loosening a net:

- Tight net → Fewer false alarms → High Precision  
- Wide net → Catch more positives → High Recall  

---

# 🏁 Final Summary

- **Precision** → How correct are positive predictions?  
- **Recall** → How many real positives were captured?  

If mistakes are expensive,  
choose the metric that matches the cost of that mistake.

# 🎯 Precision vs Recall (Simple Explanation)

Let’s forget formulas for a minute.

Imagine your model is a **security guard** at a gate checking for suspicious people.

Now two important questions:

1. When the guard stops someone, how often is he correct?
2. Did the guard miss any suspicious people?

Those two questions are **Precision** and **Recall**.

---

# 🔵 What is Precision? (Think: Accuracy of Positive Predictions)

### Simple Meaning:
When the model says **“Yes, this is positive”**,  
how often is it actually correct?

### It answers:
👉 *“Can I trust the model when it says Positive?”*

### Use Precision When:
False alarms are costly.

### Examples:
- 📧 Spam detection (important email marked as spam)
- 💳 Loan approval (bad customer approved)
- 🔎 Search results (irrelevant result shown)

You want fewer wrong positive predictions.

---

# 🟢 What is Recall? (Think: Ability to Find All Positives)

### Simple Meaning:
Out of all real positive cases,  
how many did the model successfully catch?

### It answers:
👉 *“Is the model missing important cases?”*

### Use Recall When:
Missing a positive case is dangerous.

### Examples:
- 🏥 Disease detection
- 💳 Fraud detection
- 🔐 Security threats

You don’t want the model to miss real positive cases.

---

# ⚖️ Key Difference in Simple Language

| Precision | Recall |
|------------|---------|
| How correct are positive predictions? | How many real positives were found? |
| Focuses on false positives | Focuses on false negatives |
| About prediction quality | About coverage of positives |

---

# 🎯 Quick Analogy

Imagine fishing 🎣

- **High Precision** → You catch fewer fish, but almost all are the right type.
- **High Recall** → You catch almost every fish of that type, but you may also catch some unwanted ones.

---

# 🏁 Final Understanding

- Use **Precision** when false alarms hurt.
- Use **Recall** when missing real cases hurts.
- They usually move in opposite directions.
- Balance both using **F1-score** when needed.

If accuracy tells you *how many were correct*,  
precision and recall tell you *what kind of mistakes you are making*.

# 🎯 What is F1 Score?

F1 Score is a metric that **balances Precision and Recall**.

It is used when you want both:
- Few false alarms (high Precision)
- Few missed positives (high Recall)

---

# 📌 Simple Definition

F1 Score is the **harmonic mean** of Precision and Recall.

F1 = 2 × (Precision × Recall) / (Precision + Recall)

---

# 🧠 Simple Meaning (No Math)

If:

- Precision tells you **how correct your positive predictions are**
- Recall tells you **how many real positives you captured**

Then:

👉 F1 Score tells you **how good your model is at both together**

It punishes the model if one of them is very low.

---

# 📦 Example

Suppose:

Precision = 0.90  
Recall = 0.50  

F1 will NOT be high.

F1 = 2 × (0.90 × 0.50) / (0.90 + 0.50)  
F1 = 0.64  

Even though precision is high, recall is low.  
F1 exposes that imbalance.

---

# 🎯 When Do We Use F1 Score?

Use F1 when:

- Dataset is imbalanced
- Both false positives and false negatives matter
- You need a single metric to balance Precision & Recall

### Common Use Cases:
- Fraud detection
- Disease detection
- Spam filtering

---

# ⚖️ Why Not Just Use Accuracy?

Accuracy can look high even if the model ignores the minority class.

F1 focuses only on the positive class performance  
and balances mistakes.

---

# 🧠 Quick Interview Trick

Precision → Quality  
Recall → Coverage  
F1 → Balance of Quality and Coverage  

If interviewer asks:

> Why harmonic mean?

Because harmonic mean gives more weight to lower values.  
If either precision or recall is low, F1 drops sharply.

---

# 🏁 One-Line Summary

F1 Score is the balance meter between Precision and Recall.
If one falls, F1 falls.
