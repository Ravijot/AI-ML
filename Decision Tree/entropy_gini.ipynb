## Entropy

Entropy measures the amount of uncertainty or impurity in a dataset.

$$
Entropy(S) = - \sum_{i=1}^{c} p_i \log_2(p_i)
$$

Where:

- \(S\) = dataset  
- \(c\) = number of classes  
- \(p_i\) = probability of class \(i\)

### Example (Binary Classification)

$$
Entropy(S) = -(p_1 \log_2 p_1 + p_2 \log_2 p_2)
$$

If:

- \(p_1 = 0.6\)
- \(p_2 = 0.4\)

Then:

$$
Entropy(S) = -(0.6 \log_2 0.6 + 0.4 \log_2 0.4)
$$

$$
Entropy(S) \approx 0.97
$$

**Entropy**

```python
import numpy as np
```

```python
def entropy(var):
    N = var.shape[0]
    values, counts = np.unique(var, return_counts=True )
    
    ent = 0.0
    
    for i in counts:
        p = i/N
        if p > 0:
            ent += (p * np.log2(p))
        
    return -ent
```

```python
Y = np.array([1,0,0,1,0,1,0,1])
X = np.array([1,1,1,1,1,1,1])
Z = np.array([0,0,0,0,0,0,0])
```

```python
entropy(Y) # Max randomness leads to higher entropy
```

```python
entropy(X) #No randomness leads to zero entropy
```

```python
entropy(Z) #No randomness leads to zero entropy
```

```python
import numpy as np
import matplotlib.pyplot as plt

p = np.linspace(0.001, 0.999, 100)

entropies = -(p * np.log2(p) + (1-p) * np.log2(1-p))

plt.plot(p, entropies)
plt.xlabel("Probability p")
plt.ylabel("Entropy")
plt.title("Entropy vs Probability")
plt.show()
```

## Gini Impurity

The Gini Impurity is defined as:

$$
\text{Gini}(S) = 1 - \sum_{i=1}^{c} p_i^2
$$

Where:

- \(S\) = dataset  
- \(c\) = number of classes  
- \(p_i\) = probability of class \(i\)

### Example (Binary Classification)

$$
\text{Gini}(S) = 1 - (p_1^2 + p_2^2)
$$

If:

- \(p_1 = 0.6\)
- \(p_2 = 0.4\)

Then:

$$
\text{Gini} = 1 - (0.6^2 + 0.4^2) = 0.48
$$

```python
import numpy as np

def gini(var):
    N = var.shape[0]
    values, counts = np.unique(var, return_counts=True)
    
    gini_impurity = 1.0
    
    for i in counts:
        p = i / N
        gini_impurity -= p**2
        
    return gini_impurity


Y = np.array([1,0,0,1,0,1,0,1])
X = np.array([1,1,1,1,1,1,1])
Z = np.array([0,0,0,0,0,0,0])

print("Gini(Y):", gini(Y))
print("Gini(X):", gini(X))
print("Gini(Z):", gini(Z))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# probability values
p = np.linspace(0, 1, 100)

# gini impurity formula
gini = 1 - (p**2 + (1-p)**2)

plt.plot(p, gini)
plt.xlabel("Probability p")
plt.ylabel("Gini Impurity")
plt.title("Gini Impurity vs Probability")
plt.show()
```
