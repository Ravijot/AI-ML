# 🎯 Out-of-Bag (OOB) in Bagging & Random Forest

## 🔹 What is OOB?

**Out-of-Bag (OOB)** samples are the data points **not selected** in a bootstrap sample when training a model.

👉 Since sampling is done **with replacement**, some data points are left out in each bootstrap dataset.

---

## 🔹 Key Idea

- Each tree is trained on ~**63%** of the data
- Remaining ~**37%** is **OOB data**
- These unused samples act like a **validation set**

---

## 🔹 How OOB Works

For each data point:
1. Find all trees where this point was **not used in training**
2. Use those trees to **predict** its label
3. Aggregate predictions (majority vote / average)
4. Compare with true label

👉 This gives **OOB error**

---

## 🔹 Why OOB is Useful

- No need for separate validation set ❌
- Saves data (especially useful for small datasets)
- Gives **unbiased estimate of model performance**

---

## 🔹 Where It Is Used

- ✔️ Bagging
- ✔️ Random Forest
