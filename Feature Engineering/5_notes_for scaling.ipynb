# Feature Scaling Methods – Quick Notes

## 1. Absolute Maximum Scaling
**Method**  
Divides each value by the maximum absolute value of the feature.

**Sensitivity to Outliers**  
High

**Typical Use Cases**  
- Sparse data  
- When preserving zero values is important  
- Simple, fast scaling

---

## 2. Min–Max Scaling (Normalization)
**Method**  
Scales features to a fixed range, usually [0, 1], using min–max normalization.

**Sensitivity to Outliers**  
High

**Typical Use Cases**  
- Neural networks  
- Models requiring bounded inputs  
- Image and pixel data

---

## 3. Normalization (Vector Norm)
**Method**  
Scales each sample (row) to have unit norm (length = 1).

**Sensitivity to Outliers**  
Not applicable (operates per row, not per feature)

**Typical Use Cases**  
- Direction-based similarity  
- Text classification (TF-IDF, cosine similarity)

---

## 4. Standardization (Z-Score Scaling)
**Method**  
Centers features to mean = 0 and scales to unit variance (std = 1).

**Sensitivity to Outliers**  
Moderate

**Typical Use Cases**  
- Most machine learning algorithms  
- Logistic Regression, SVM, PCA  
- Data approximately normally distributed

---

## 5. Robust Scaling
**Method**  
Centers data using the median and scales using the interquartile range (IQR).

**Sensitivity to Outliers**  
Low

**Typical Use Cases**  
- Data with outliers  
- Skewed distributions  
- Real-world noisy datasets

---

## Key Takeaway
- Use **Standardization** by default  
- Use **Min–Max** when bounds matter  
- Use **Robust Scaling** when outliers dominate  
- Use **Normalization** for direction-based similarity


# Feature Scaling – Algorithm Requirements

## Why Scaling Matters
Feature scaling is required when an algorithm:
- Uses **distance calculations**
- Uses **dot products**
- Uses **gradient-based optimization**

Tree-based models do **not** rely on distances, so scaling does not affect them.

---

## Algorithms That NEED Scaling ✅

### 1. Distance-Based Algorithms
Scaling is **mandatory**.

- K-Nearest Neighbors (KNN)
- K-Means Clustering
- DBSCAN
- Hierarchical Clustering

**Reason:**  
Distance metrics (Euclidean, Manhattan) are affected by feature magnitude.

---

### 2. Linear & Gradient-Based Models
Scaling is **strongly recommended**.

- Linear Regression
- Logistic Regression
- Ridge / Lasso / ElasticNet
- Support Vector Machines (SVM)
- Perceptron
- Neural Networks

**Reason:**  
- Faster convergence  
- Stable gradient descent  
- Coefficients become comparable  

---

### 3. Dimensionality Reduction Techniques
Scaling is **mandatory**.

- Principal Component Analysis (PCA)
- Linear Discriminant Analysis (LDA)
- t-SNE
- UMAP

**Reason:**  
Variance and distances must be comparable across features.

---

## Algorithms That DO NOT Need Scaling ❌

### Tree-Based Models
Scaling has **no impact** on performance.

- Decision Trees
- Random Forest
- Gradient Boosting
- XGBoost
- LightGBM
- CatBoost

**Reason:**  
Trees split on feature thresholds, not distances.
