**One Hot Encoding**  
One-hot encoding is a data preprocessing technique that converts categorical data (like 'Red', 'Green', 'Blue') into a numerical format for machine learning models by creating new binary columns for each unique category, assigning a '1' in the corresponding column and '0' in others, preventing models from misinterpreting categories as having an order or magnitude. 

**How it works**  
Identifies Categories: Finds all unique values (e.g., 'Apple', 'Banana', 'Mango') in a categorical column.  
Creates New Columns: Generates a new binary column for each unique category.  
Assigns Binary Values: For each original data point, places a '1' in the new column corresponding to its category and '0' in all other new columns.

**Why it's used**
Numerical Input: Most ML algorithms require numerical input.  
No False Order: Prevents models from assuming an order (e.g., 'Apple' < 'Banana') that doesn't exist, which can happen with simple integer encoding.  
Handles Nominal Data: Ideal for nominal (unranked) categorical data like colors, cities, or types.  

**Key Considerations**
Curse of Dimensionality: Can create too many features (columns) if a category has many unique values, potentially slowing models.  
Multicollinearity: Introduces dependency between the new columns, though often handled by dropping one category (e.g., if it's not Apple or Banana, it must be Mango).

Example  
If you have a 'Fruit' column with 'Apple', 'Banana', 'Mango':  
Original: Apple, Banana, Mango  
One-Hot Encoded:    
Apple becomes  (Apple=1, Banana=0, Mango=0)    
Banana becomes (Apple=0, Banana=1, Mango=0)    
Mango becomes  (Apple=0, Banana=0, Mango=1)

**OneHotEncoding using Sklearn**

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,0:4],df.iloc[:,-1],test_size=0.2,random_state=2)
```

```python
ohe = OneHotEncoder(drop='first',dtype=np.int32)
```

```python
X_test_new = ohe.transform(X_test[['fuel','owner']])
```
